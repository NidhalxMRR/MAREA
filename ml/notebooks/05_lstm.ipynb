# 05 LSTM

## Purpose

Train and evaluate an LSTM neural network for temperature forecasting:

- **Only proceed if baselines justify deep learning**
  - LSTM must demonstrate > 10% improvement over feature model to warrant complexity

- **Architecture Design**:
  - Lookback window (e.g., 48 steps = 12 hours of 15-min data)
  - LSTM layers + dropout for regularization
  - Dense output layer for each forecast horizon

- **Training**:
  - Chronological train/validation/test split
  - Fit scaler on training data only
  - Early stopping on validation set
  - Save model checkpoints

**Evaluation**:
- Compare LSTM against baselines and feature model
- Record MAE, RMSE, R² for each forecast horizon
- Analyze predictions: error distribution, worst-case scenarios
- Decide: deploy or continue iteration

**Output**: LSTM model saved to `../models/temperature_lstm_v*.keras` with metadata JSON
