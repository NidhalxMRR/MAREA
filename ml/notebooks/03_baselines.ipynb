# 03 Baselines

## Purpose

Establish baseline forecasting performance before attempting sophisticated models:

- **Persistence Baseline**: Next temperature = most recent observed temperature
  - Trivial lower bound for any forecast
  - Always required as sanity check

- **Moving Average / Seasonal Baseline**: Simple statistical forecasting
  - Rolling mean over recent history
  - Captures basic temporal patterns

**Evaluation**:
- Chronological train/validation/test split
- Compute MAE, RMSE, R² on test set
- Record baseline scores for comparison with feature models and LSTM

**Output**: Baseline metrics saved to `../reports/metrics/03_baselines_results.json`
