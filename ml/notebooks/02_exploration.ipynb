# 02 Exploration

## Purpose

Analyze temporal patterns and statistical properties of the validated dataset:

- **Distributions**: Histogram, quantiles, outlier detection
- **Seasonality**: Daily, weekly, annual patterns
- **Autocorrelation**: Temporal correlation at various lags
- **Rate of Change**: Temperature change per timestep
- **Anomaly Behavior**: How flagged anomalies relate to normal data
- **Stationarity**: Time-series decomposition

**Output**: Exploratory plots saved to `../reports/figures/02_exploration_*.png`
