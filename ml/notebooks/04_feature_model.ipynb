# 04 Feature Model

## Purpose

Develop and evaluate a feature-based forecasting model:

- **Feature Engineering**: Create lag features, rolling statistics
  - Lag features: temperatures at t-1, t-4, t-24, etc.
  - Rolling mean / std over windows
  - Rate of change and other domain features

- **Model Selection**: Use interpretable ML
  - Linear regression, decision trees, random forest, XGBoost (optional)
  - Do not select library yet—compare options

- **Training**: Chronological train/validation/test split
  - Fit feature statistics and model on training data only
  - Evaluate on held-out test set

**Evaluation**:
- Compare feature model against baselines
- Record MAE, RMSE, R² for each forecast horizon
- Assess feature importance / interpretability

**Output**: Model performance and feature importance saved to `../reports/metrics/04_feature_model_results.json`
