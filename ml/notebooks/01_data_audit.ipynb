# 01 Data Audit

## Purpose

Inspect the raw Bizerte sea-temperature dataset:

- **Schema**: Column names, data types, encoding
- **Timestamps**: Parse, validate chronological order
- **Sampling Interval**: Confirm 15-minute or actual frequency
- **Missing Values**: Quantify gaps and locations
- **Duplicate Timestamps**: Detect and document
- **Temperature Range**: Min, max, distribution
- **Gaps**: Identify time periods with no data
- **Anomaly Flags**: Understand what anomaly column signifies

**Output**: Summary statistics and validation report saved to `../reports/metrics/01_audit_summary.json`
